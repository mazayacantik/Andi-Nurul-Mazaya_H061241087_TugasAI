# PROMPT 1
# Analisis Aktivitas Gempa Bumi Berdasarkan Data Seismik

## R (Role)
Anda adalah seorang ahli geofisika bidang seismologi yang berpengalaman dalam menganalisis aktivitas gempa bumi dan kondisi tektonik suatu wilayah.

## T (Task)
Lakukan analisis terhadap data gempa bumi untuk mengetahui karakteristik aktivitas seismik, pola persebaran gempa, dan kondisi tektonik wilayah penelitian.

## C (Context)
Suatu wilayah berada pada zona tektonik aktif yang sering mengalami kejadian gempa bumi. Analisis aktivitas gempa diperlukan untuk memahami hubungan antara parameter gempa dengan kondisi geologi wilayah tersebut.

## I (Input)
Data yang tersedia:
- Lokasi episenter gempa.
- Magnitudo gempa.
- Kedalaman hiposenter.
- Waktu kejadian gempa.
- Distribusi kejadian gempa.

## C (Constraints)
- Gunakan prinsip dasar seismologi dan tektonik lempeng.
- Jangan memberikan kesimpulan tanpa berdasarkan data.
- Bedakan hasil analisis dengan interpretasi.
- Gunakan istilah ilmiah bidang geofisika.

## O (Output)
Buat hasil analisis berupa:
- Tabel karakteristik gempa.
- Analisis hubungan magnitudo dan kedalaman.
- Interpretasi pola aktivitas seismik.
- Kesimpulan kondisi tektonik wilayah.

## Q (Quality Criteria)
Jawaban harus:
- Sistematis.
- Akurat secara konsep geofisika.
- Menggunakan bahasa ilmiah yang mudah dipahami.

## Rencana Output:
Output akhir berupa laporan analisis aktivitas gempa dalam bentuk tabel, penjelasan ilmiah, dan kesimpulan kondisi seismik wilayah.


---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PROMPT 1: DATA GEMPA SINTETIS (untuk demonstrasi, bukan data observasi nyata)
np.random.seed(21)
n = 80
waktu = pd.date_range("2025-01-01", periods=n, freq="12h")
lintang = -5.5 + np.random.uniform(-1.2, 1.2, n)
bujur = 119.5 + np.random.uniform(-1.5, 1.5, n)
magnitudo = np.clip(np.random.normal(3.4, 0.7, n), 1.5, 5.8)
kedalaman = np.clip(np.random.gamma(shape=2.2, scale=13, size=n), 2, 100)

gempa = pd.DataFrame({
    "Waktu": waktu, "Lintang": lintang, "Bujur": bujur,
    "Magnitudo": magnitudo.round(2), "Kedalaman_km": kedalaman.round(1)
})
display(gempa.head(10))

# Ringkasan statistik karakteristik gempa
display(gempa[["Magnitudo", "Kedalaman_km"]].describe().round(2))

# Visualisasi hubungan magnitudo dan kedalaman
plt.figure(figsize=(8, 5))
plt.scatter(gempa["Kedalaman_km"], gempa["Magnitudo"], alpha=0.75)
plt.xlabel("Kedalaman hiposenter (km)")
plt.ylabel("Magnitudo")
plt.title("Hubungan Magnitudo dan Kedalaman Gempa (Data Sintetis)")
plt.grid(True, alpha=0.3)
plt.show()

# Persebaran episenter secara sederhana
plt.figure(figsize=(7, 5))
plt.scatter(gempa["Bujur"], gempa["Lintang"], s=gempa["Magnitudo"]**2 * 5,
            c=gempa["Kedalaman_km"], cmap="viridis", alpha=0.75)
plt.colorbar(label="Kedalaman (km)")
plt.xlabel("Bujur")
plt.ylabel("Lintang")
plt.title("Persebaran Episenter Gempa (Data Sintetis)")
plt.grid(True, alpha=0.3)
plt.show()

print("Interpretasi awal:")
print(f"- Rata-rata magnitudo: {gempa['Magnitudo'].mean():.2f}.")
print(f"- Rata-rata kedalaman: {gempa['Kedalaman_km'].mean():.1f} km.")
print("- Pola hubungan magnitudo-kedalaman perlu dinilai dari sebaran data; grafik ini saja tidak cukup untuk menetapkan mekanisme tektonik.")
print("- Data ini sintetis. Interpretasi tektonik nyata memerlukan katalog gempa, lokasi, mekanisme fokus, dan informasi geologi yang tervalidasi.")
```

# PROMPT 2
# Identifikasi Potensi Air Tanah Menggunakan Metode Geolistrik

## O (Output)

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PROMPT 2: MODEL RESISTIVITAS SINTETIS (Ohm m)
# Nilai hanya contoh konseptual; rentang resistivitas nyata sangat bervariasi.
np.random.seed(22)
x = np.linspace(0, 200, 121)       # jarak lintasan (m)
z = np.linspace(0, 60, 61)         # kedalaman (m)
X, Z = np.meshgrid(x, z)

# Latar belakang berlapis: tanah/lempung relatif konduktif,
# lapisan lebih resistif, dan zona konduktif yang dimodelkan sebagai kandidat jenuh air.
R = np.where(Z < 8, 35.0, np.where(Z < 25, 120.0, 65.0))
zona_akuifer = ((X - 112)**2 / 38**2 + (Z - 34)**2 / 12**2) < 1
R[zona_akuifer] = 22.0
R += np.random.normal(0, 3, R.shape)
R = np.clip(R, 5, None)

plt.figure(figsize=(10, 5))
kontur = plt.contourf(X, Z, R, levels=20, cmap="viridis")
plt.gca().invert_yaxis()
plt.colorbar(kontur, label="Resistivitas (Ohm m)")
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Kedalaman (m)")
plt.title("Penampang Resistivitas 2D Sintetis")
plt.show()

# Ringkasan nilai resistivitas berdasarkan interval kedalaman
baris = []
for batas_bawah, batas_atas, nama in [(0, 8, "Lapisan dangkal"),
                                      (8, 25, "Lapisan menengah"),
                                      (25, 60, "Lapisan dalam")]:
    nilai = R[(Z >= batas_bawah) & (Z < batas_atas)]
    baris.append({
        "Interval": nama,
        "Kedalaman (m)": f"{batas_bawah}-{batas_atas}",
        "Resistivitas median (Ohm m)": round(float(np.median(nilai)), 1)
    })
tabel_res = pd.DataFrame(baris)
display(tabel_res)

print("Interpretasi awal:")
print("- Zona resistivitas rendah pada model ditandai sebagai kandidat material lebih konduktif, yang dapat berkaitan dengan lempung atau material jenuh air.")
print("- Resistivitas rendah tidak otomatis membuktikan akuifer; lempung dan air dengan kandungan ion tinggi juga dapat menghasilkan nilai rendah.")
print("- Penentuan akuifer perlu mempertimbangkan konfigurasi elektroda, kualitas inversi, geologi setempat, data sumur, dan validasi lapangan.")
```

## O (Output)
Hasil berupa:

## Rencana Output:

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PROMPT 3: PENILAIAN AWAL FAKTOR LONGSOR DENGAN DATA SINTETIS
np.random.seed(23)
n = 50
lereng = np.random.uniform(5, 45, n)              # derajat
hujan = np.random.uniform(20, 220, n)             # mm/hari
kejenuhan = np.random.uniform(20, 100, n)         # persen
gangguan_lahan = np.random.uniform(0, 1, n)       # indeks 0-1

# Skor sederhana untuk latihan, bukan model resmi pemetaan bahaya
skor = (lereng / 45 * 0.35 +
        hujan / 220 * 0.30 +
        kejenuhan / 100 * 0.25 +
        gangguan_lahan * 0.10)
kelas = pd.cut(skor, bins=[-0.01, 0.33, 0.66, 1.01],
               labels=["Rendah", "Sedang", "Tinggi"])
data_longsor = pd.DataFrame({
    "Kemiringan (derajat)": lereng.round(1),
    "Curah hujan (mm/hari)": hujan.round(1),
    "Kejenuhan tanah (%)": kejenuhan.round(1),
    "Gangguan lahan (indeks)": gangguan_lahan.round(2),
    "Skor indeks": skor.round(3),
    "Kelas indeks": kelas.astype(str)
})
display(data_longsor.head(10))
display(data_longsor["Kelas indeks"].value_counts().rename_axis("Kelas").to_frame("Jumlah titik"))

plt.figure(figsize=(8, 5))
for label in ["Rendah", "Sedang", "Tinggi"]:
    subset = data_longsor[data_longsor["Kelas indeks"] == label]
    plt.scatter(subset["Kemiringan (derajat)"], subset["Curah hujan (mm/hari)"],
                label=label, alpha=0.75)
plt.xlabel("Kemiringan lereng (derajat)")
plt.ylabel("Curah hujan (mm/hari)")
plt.title("Sebaran Faktor Lereng dan Hujan Berdasarkan Indeks Sintetis")
plt.legend(title="Kelas indeks")
plt.grid(True, alpha=0.3)
plt.show()

print("Interpretasi awal:")
print("- Skor dibuat dari kombinasi kemiringan, hujan, kejenuhan, dan gangguan lahan dengan bobot ilustratif.")
print("- Titik dengan lereng lebih curam dan hujan lebih tinggi cenderung memperoleh skor lebih besar pada model ini.")
print("- Kelas pada notebook ini bukan peta resmi kerawanan longsor. Analisis nyata membutuhkan inventaris kejadian longsor, litologi, struktur geologi, faktor keamanan lereng, resolusi spasial, dan validasi lapangan.")
```

# Interpretasi Data Magnetik untuk Eksplorasi Mineral

## R (Role)
Anda adalah seorang ahli geofisika eksplorasi yang menguasai metode geomagnetik untuk penelitian sumber daya mineral.

## T (Task)
Analisis data magnetik untuk mengetahui pola anomali dan kemungkinan struktur geologi bawah permukaan.

Suatu wilayah memiliki indikasi keberadaan mineral sehingga diperlukan metode geofisika untuk mengetahui kondisi bawah permukaan sebelum eksplorasi lebih lanjut.

## I (Input)

## C (Constraints)

## O (Output)
Hasil analisis berupa:

## Q (Quality Criteria)

## Rencana Output:
Output berupa tabel anomali magnetik, interpretasi struktur bawah permukaan, dan analisis potensi mineral.


---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PROMPT 4: DATA ANOMALI MAGNETIK SINTETIS (nT)
np.random.seed(24)
x = np.linspace(0, 1000, 250)  # jarak lintasan (m)
regional = 12 * np.sin(x / 260)
anomali_positif = 95 * np.exp(-((x - 280)**2) / (2 * 65**2))
anomali_negatif = -70 * np.exp(-((x - 690)**2) / (2 * 90**2))
anomali_lokal = 45 * np.exp(-((x - 850)**2) / (2 * 35**2))
noise = np.random.normal(0, 5, x.size)
total = regional + anomali_positif + anomali_negatif + anomali_lokal + noise

mag = pd.DataFrame({
    "Jarak (m)": x.round(1),
    "Anomali magnetik (nT)": total.round(2)
})
display(mag.head(10))
display(mag["Anomali magnetik (nT)"].describe().to_frame("Statistik").round(2))

plt.figure(figsize=(10, 5))
plt.plot(x, total, label="Anomali total", linewidth=1.5)
plt.plot(x, regional, label="Komponen regional model", linestyle="--")
plt.axhline(0, linewidth=0.8)
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Anomali magnetik (nT)")
plt.title("Profil Anomali Magnetik Sintetis")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Menandai nilai anomali ekstrem secara statistik untuk pemeriksaan awal
mean_mag = mag["Anomali magnetik (nT)"].mean()
std_mag = mag["Anomali magnetik (nT)"].std()
mag["Kategori awal"] = np.where(mag["Anomali magnetik (nT)"] > mean_mag + std_mag, "Relatif tinggi",
                         np.where(mag["Anomali magnetik (nT)"] < mean_mag - std_mag, "Relatif rendah", "Dekat kisaran umum"))
display(mag[mag["Kategori awal"] != "Dekat kisaran umum"].head(12))

print("Interpretasi awal:")
print("- Puncak dan lembah pada profil dapat mencerminkan variasi kemagnetan batuan, geometri sumber, atau efek pemrosesan.")
print("- Anomali magnetik tidak secara langsung membuktikan keberadaan mineral ekonomis.")
print("- Interpretasi memerlukan koreksi harian/IGRF yang sesuai, informasi geologi, pemisahan regional-residual, serta survei pendukung.")
```

# PROMPT 5
# Monitoring Aktivitas Gunung Api Berdasarkan Data Geofisika

## R (Role)
Anda adalah seorang ahli vulkanologi dan geofisika yang melakukan pemantauan aktivitas gunung api.

## T (Task)
Analisis perubahan aktivitas gunung api berdasarkan data seismik, deformasi, dan parameter vulkanologi.

## C (Context)
Gunung api aktif membutuhkan pemantauan secara berkala untuk mengetahui perubahan aktivitas magma dan peningkatan aktivitas vulkanik.

## I (Input)
Data yang tersedia:
- Jumlah gempa vulkanik.
- Data deformasi permukaan.
- Perubahan suhu kawah.
- Aktivitas visual gunung api.

## C (Constraints)
- Gunakan prinsip pemantauan gunung api modern.
- Jangan memberikan prediksi erupsi secara pasti.
- Gunakan data sebagai dasar interpretasi.

## O (Output)
Hasil analisis berupa:
- Perubahan parameter aktivitas gunung api.
- Interpretasi kondisi magma.
- Indikator peningkatan aktivitas.
- Rekomendasi pemantauan.

## Q (Quality Criteria)
Jawaban harus:
- Ilmiah.
- Sistematis.

- Menjelaskan hubungan antarparameter.

## V (Verification)
Bandingkan hasil analisis dengan pola aktivitas gunung api berdasarkan data historis dan referensi vulkanologi.

## Rencana Output:
Output berupa tabel perubahan aktivitas, interpretasi kondisi gunung api, dan rekomendasi pemantauan lanjutan.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PROMPT 5: PARAMETER PEMANTAUAN GUNUNG API SINTETIS
# Data berikut hanya simulasi untuk latihan analisis tren.
np.random.seed(25)
hari = np.arange(1, 31)
gempa_vulkanik = np.random.poisson(lam=8, size=30)
# Simulasikan kenaikan bertahap pada bagian akhir bulan
gempa_vulkanik[20:] += np.arange(0, 10)
deformasi_mm = np.cumsum(np.random.normal(0.05, 0.18, 30))
suhu_kawah_c = 78 + 0.12 * hari + np.random.normal(0, 1.4, 30)
visual_index = np.clip(2 + np.random.normal(0, 0.45, 30) + (hari > 22) * 0.7, 0, 5)

gunung = pd.DataFrame({
    "Hari": hari,
    "Gempa vulkanik (kejadian/hari)": gempa_vulkanik,
    "Deformasi kumulatif (mm)": deformasi_mm.round(2),
    "Suhu kawah (°C)": suhu_kawah_c.round(1),
    "Indeks visual (0-5)": visual_index.round(1)
})
display(gunung.head(10))
display(gunung[["Gempa vulkanik (kejadian/hari)", "Deformasi kumulatif (mm)",
                "Suhu kawah (°C)", "Indeks visual (0-5)"]].describe().round(2))

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].plot(hari, gempa_vulkanik, marker="o")
axes[0, 0].set_title("Jumlah Gempa Vulkanik")
axes[0, 0].set_xlabel("Hari"); axes[0, 0].set_ylabel("Kejadian/hari")
axes[0, 1].plot(hari, deformasi_mm, marker="o")
axes[0, 1].set_title("Deformasi Kumulatif")
axes[0, 1].set_xlabel("Hari"); axes[0, 1].set_ylabel("mm")
axes[1, 0].plot(hari, suhu_kawah_c, marker="o")
axes[1, 0].set_title("Suhu Kawah")
axes[1, 0].set_xlabel("Hari"); axes[1, 0].set_ylabel("°C")
axes[1, 1].plot(hari, visual_index, marker="o")
axes[1, 1].set_title("Indeks Pengamatan Visual")
axes[1, 1].set_xlabel("Hari"); axes[1, 1].set_ylabel("Indeks 0-5")
for ax in axes.flat:
    ax.grid(True, alpha=0.3)
plt.suptitle("Tren Parameter Pemantauan Gunung Api (Data Sintetis)")
plt.tight_layout()
plt.show()

# Perbandingan rata-rata awal dan akhir periode
awal = gunung.iloc[:10]
akhir = gunung.iloc[-10:]
ringkasan = pd.DataFrame({
    "Parameter": ["Gempa vulkanik (kejadian/hari)", "Deformasi kumulatif (mm)",
                  "Suhu kawah (°C)", "Indeks visual (0-5)"],
    "Rata-rata 10 hari awal": [
        awal["Gempa vulkanik (kejadian/hari)"].mean(),
        awal["Deformasi kumulatif (mm)"].mean(),
        awal["Suhu kawah (°C)"].mean(),
        awal["Indeks visual (0-5)"].mean()
    ],
    "Rata-rata 10 hari akhir": [
        akhir["Gempa vulkanik (kejadian/hari)"].mean(),
        akhir["Deformasi kumulatif (mm)"].mean(),
        akhir["Suhu kawah (°C)"].mean(),
        akhir["Indeks visual (0-5)"].mean()
    ]
})
display(ringkasan.round(2))

print("Interpretasi awal:")
print("- Perubahan beberapa parameter secara bersamaan dapat menjadi alasan untuk meningkatkan evaluasi dan pemantauan.")
print("- Tren sintetis tidak dapat digunakan untuk menentukan status gunung api atau memprediksi waktu erupsi.")
print("- Penilaian operasional harus mengikuti data resmi, prosedur lembaga pemantauan gunung api, dan evaluasi ahli.")
```
