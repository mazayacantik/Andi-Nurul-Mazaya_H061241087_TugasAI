# PROMPT 1
# Analisis Aktivitas Gempa Bumi Berdasarkan Data Seismik

## R (Role)
Anda adalah seorang ahli geofisika bidang seismologi yang berpengalaman dalam menganalisis aktivitas gempa bumi dan kondisi tektonik suatu wilayah.

## T (Task)
Lakukan analisis terhadap data gempa bumi untuk mengetahui karakteristik aktivitas seismik, pola persebaran gempa, dan kondisi tektonik wilayah penelitian.

## C (Context)
Suatu wilayah berada pada zona tektonik aktif yang sering mengalami kejadian gempa bumi. Analisis aktivitas gempa diperlukan untuk memahami hubungan antara parameter gempa dengan kondisi geologi wilayah tersebut.

## I (Input)
Data yang tersedia:
- Lokasi episenter gempa.
- Magnitudo gempa.
- Kedalaman hiposenter.
- Waktu kejadian gempa.
- Distribusi kejadian gempa.

## C (Constraints)
- Gunakan prinsip dasar seismologi dan tektonik lempeng.
- Jangan memberikan kesimpulan tanpa berdasarkan data.
- Bedakan hasil analisis dengan interpretasi.
- Gunakan istilah ilmiah bidang geofisika.

## O (Output)
Buat hasil analisis berupa:
- Tabel karakteristik gempa.
- Analisis hubungan magnitudo dan kedalaman.
- Interpretasi pola aktivitas seismik.
- Kesimpulan kondisi tektonik wilayah.

## Q (Quality Criteria)
Jawaban harus:
- Sistematis.
- Akurat secara konsep geofisika.
- Menggunakan bahasa ilmiah yang mudah dipahami.
- Memiliki hubungan jelas antara data dan hasil analisis.

## V (Verification)
Verifikasi hasil dengan membandingkan interpretasi aktivitas gempa terhadap teori tektonik lempeng dan penelitian terdahulu.

## Rencana Output:
Output akhir berupa laporan analisis aktivitas gempa dalam bentuk tabel, penjelasan ilmiah, dan kesimpulan kondisi seismik wilayah.

# PROMPT 2
# Identifikasi Potensi Air Tanah Menggunakan Metode Geolistrik

## R (Role)
Anda adalah seorang ahli geofisika eksplorasi yang memiliki keahlian dalam metode resistivitas listrik untuk kajian bawah permukaan.

## T (Task)
Analisis data geolistrik untuk mengetahui struktur lapisan bawah permukaan dan menentukan kemungkinan keberadaan akuifer.

## C (Context)
Sebuah daerah mengalami keterbatasan sumber air sehingga diperlukan penelitian geofisika untuk mengetahui kondisi bawah permukaan tanpa melakukan pengeboran awal.

## I (Input)
Data yang diberikan:
- Nilai resistivitas batuan.
- Konfigurasi elektroda.
- Kedalaman investigasi.
- Data hasil pengukuran lapangan.

## C (Constraints)
- Gunakan hubungan nilai resistivitas dengan karakteristik material bumi.
- Jangan menentukan jenis batuan hanya berdasarkan satu parameter.
- Gunakan konsep geofisika dan hidrogeologi.

## O (Output)
Hasil analisis berupa:
- Tabel interpretasi nilai resistivitas.
- Identifikasi lapisan bawah permukaan.
- Penentuan zona potensial air tanah.
- Model konseptual struktur bawah tanah.

## Q (Quality Criteria)
Jawaban harus:
- Berdasarkan prinsip metode geolistrik.
- Memiliki alasan ilmiah.
- Menjelaskan keterbatasan interpretasi.

## V (Verification)
Bandingkan hasil interpretasi dengan nilai resistivitas referensi material geologi dan kondisi lapangan.

## Rencana Output:
Output berupa tabel resistivitas, interpretasi lapisan bawah permukaan, dan kesimpulan potensi akuifer.

# PROMPT 3
# Analisis Risiko Longsor Berdasarkan Data Geofisika

## R (Role)
Anda adalah seorang ahli geofisika lingkungan yang melakukan kajian mitigasi bencana longsor.

## T (Task)
Analisis tingkat kerawanan longsor suatu wilayah berdasarkan faktor geologi, kondisi lereng, dan data geofisika.

## C (Context)
Wilayah penelitian memiliki kondisi lereng yang curam dan sering mengalami kejadian longsor akibat faktor alam serta perubahan penggunaan lahan.

## I (Input)
Data yang tersedia:
- Kemiringan lereng.
- Jenis batuan.
- Kondisi tanah.
- Curah hujan.
- Data survei geofisika bawah permukaan.

## C (Constraints)
- Gunakan pendekatan geologi dan geofisika.
- Jangan menentukan tingkat bahaya hanya berdasarkan satu faktor.
- Jelaskan hubungan antarparameter yang digunakan.

## O (Output)
Hasil berupa:
- Faktor penyebab longsor.
- Klasifikasi tingkat kerawanan.
- Interpretasi kondisi bawah permukaan.
- Rekomendasi mitigasi.

## Q (Quality Criteria)
Jawaban harus:
- Logis.
- Sistematis.
- Sesuai konsep kestabilan lereng.
- Menggunakan istilah geofisika yang tepat.

## V (Verification)
Validasi hasil analisis dengan teori kestabilan lereng dan penelitian mitigasi longsor sebelumnya.

## Rencana Output:
Output berupa matriks faktor longsor, klasifikasi kerawanan wilayah, dan rekomendasi mitigasi.

# PROMPT 4
# Interpretasi Data Magnetik untuk Eksplorasi Mineral

## R (Role)
Anda adalah seorang ahli geofisika eksplorasi yang menguasai metode geomagnetik untuk penelitian sumber daya mineral.

## T (Task)
Analisis data magnetik untuk mengetahui pola anomali dan kemungkinan struktur geologi bawah permukaan.

## C (Context)
Suatu wilayah memiliki indikasi keberadaan mineral sehingga diperlukan metode geofisika untuk mengetahui kondisi bawah permukaan sebelum eksplorasi lebih lanjut.

## I (Input)
Data yang tersedia:
- Nilai anomali magnetik.
- Intensitas medan magnet.
- Koordinat pengukuran.
- Informasi geologi wilayah.

## C (Constraints)
- Gunakan konsep metode geomagnetik.
- Jangan memastikan keberadaan mineral hanya berdasarkan data magnetik.
- Jelaskan tingkat ketidakpastian interpretasi.

## O (Output)
Hasil analisis berupa:
- Identifikasi pola anomali magnetik.
- Interpretasi struktur bawah permukaan.
- Hubungan anomali dengan batuan/mineral.
- Kesimpulan potensi eksplorasi.

## Q (Quality Criteria)
Jawaban harus:
- Berdasarkan prinsip geofisika eksplorasi.
- Memiliki penjelasan ilmiah.
- Tidak memberikan kesimpulan berlebihan.

## V (Verification)
Verifikasi hasil dengan membandingkan interpretasi data magnetik terhadap informasi geologi permukaan dan penelitian sebelumnya.

## Rencana Output:
Output berupa tabel anomali magnetik, interpretasi struktur bawah permukaan, dan analisis potensi mineral.

# PROMPT 5
# Monitoring Aktivitas Gunung Api Berdasarkan Data Geofisika

## R (Role)
Anda adalah seorang ahli vulkanologi dan geofisika yang melakukan pemantauan aktivitas gunung api.

## T (Task)
Analisis perubahan aktivitas gunung api berdasarkan data seismik, deformasi, dan parameter vulkanologi.

## C (Context)
Gunung api aktif membutuhkan pemantauan secara berkala untuk mengetahui perubahan aktivitas magma dan peningkatan aktivitas vulkanik.

## I (Input)
Data yang tersedia:
- Jumlah gempa vulkanik.
- Data deformasi permukaan.
- Perubahan suhu kawah.
- Aktivitas visual gunung api.

## C (Constraints)
- Gunakan prinsip pemantauan gunung api modern.
- Jangan memberikan prediksi erupsi secara pasti.
- Gunakan data sebagai dasar interpretasi.

## O (Output)
Hasil analisis berupa:
- Perubahan parameter aktivitas gunung api.
- Interpretasi kondisi magma.
- Indikator peningkatan aktivitas.
- Rekomendasi pemantauan.

## Q (Quality Criteria)
Jawaban harus:
- Ilmiah.
- Sistematis.
- Menggunakan konsep vulkanologi.
- Menjelaskan hubungan antarparameter.

## V (Verification)
Bandingkan hasil analisis dengan pola aktivitas gunung api berdasarkan data historis dan referensi vulkanologi.

## Rencana Output:
Output berupa tabel perubahan aktivitas, interpretasi kondisi gunung api, dan rekomendasi pemantauan lanjutan.
